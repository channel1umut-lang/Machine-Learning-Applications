```python
X=df.drop('subscribed',axis=1)
y=df['subscribed']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.30,random_state=15) #random state = 15: ordaki rakamin bir degeri yok sadece ayni dagilimi herseferinde tekrar kullanmak icin
```

```python
from sklearn.linear_model import LogisticRegression
logistic=LogisticRegression()
logistic.fit(X_train,y_train)
y_pred=logistic.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
```

```python
score=accuracy_score(y_pred,y_test)
print("score: " , score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n" , confusion_matrix(y_pred,y_test))
```

```python
#hyperparameter tuning----> yeni parametreleri deniyor. amac daha iyi parametrelerle accuracy oraninlarini vs iyilestirmek
```

```python
model=LogisticRegression()
penalty=['l1', 'l2', 'elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg', 'lbfgs', 'liblinear', 'sag', 'saga']
```

```python
params=dict(penalty=penalty,C=c_values,solver=solver)
```

```python
params
```

```python
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold()
```

```python
## GridSearchCV----> bizim girdigimiz parametrelerden en iyi dogruluk oranina sahip olani deneyerek bulup bize gosteriyor.
from sklearn.model_selection import GridSearchCV

grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1) # estimetor= yukardaki model degiskeni, param_grid = deneyecegi parametreler, scoring = accuracy mi? F1 mi? recall mi? bunu belirliyor
                                                                                        # n_jobs=-1 cpu daki butun islemcileri kullan demek
```

```python
grid
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_ # best parametreleri gosteriyor
```

```python
grid.best_score_
```

```python
y_pred=grid.predict(X_test)
```

```python
score=accuracy_score(y_pred,y_test)
print("score: " , score)
print(classification_report(y_pred,y_test))
print("confusion matrix: \n" , confusion_matrix(y_pred,y_test))
```
