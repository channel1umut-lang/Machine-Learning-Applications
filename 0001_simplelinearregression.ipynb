```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
mse = mean_squared_error(y_test, y_pred_test) # MSE = (1/n) * Σ(yᵢ - ŷᵢ)²
mae = mean_absolute_error(y_test, y_pred_test) # MAE = (1/n) * Σ|yᵢ - ŷᵢ|----ŷᵢ = i'nci tahmin edilen değer---- yᵢ = i'nci gerçek (gerçekte olan) değer--n = Veri sayısı (gözlem sayısı)
rmse = np.sqrt(mse) #
print("mse:", mse)
print("mae:", mae)
print("rmse:", rmse)
```

```python
#r2---------R² = 1 - [ Σ(yᵢ - ŷᵢ)² / Σ(yᵢ - ȳ)² ] yᵢ = Gerçek (gözlenen) değerler ŷᵢ = Tahmin edilen değerler ȳ = Gerçek değerlerin ortalaması Σ = Toplam sembolü (i=1'den n'e kadar)
# Kareler Toplamı (Hata) = Σ(yᵢ - ŷᵢ)² ----gercek eksi tahmin
# Genel Kareler Toplamı = Σ(yᵢ - ȳ)²---gercek eksi ortalama
r2 = r2_score(y_test, y_pred_test)
print("r2:", r2)
```

```python
#adj r2 score #from google
```

```python
r2_adj = 1-(1-r2)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
```
