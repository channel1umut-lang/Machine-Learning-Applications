```python
X = df.drop("attack_type", axis=1)
y = df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: \n ", confusion_matrix(y_pred, y_test))
```

```python
penalty = ["l1", "l2", "elasticnet"]  # ceza metriklerini ayarliyoruz
c_values = [100, 10, 1, 0.1, 0.01]  # ne kadar guclu bir ceza verecegimizi temsil ediyor. lamda orani
solver = ["newton-cg", "lbfgs", "liblinear", "sag", "saga", "newton-cholesky"]#Yani, modelin katsayılarını (weights) bulmak için hangi algoritmanın kullanılacağını belirliyorsun. Her birinin avantajları ve kullanım alanları biraz farklıdır. L1 L2 vs yi temsil ediyor liblinear vs.
```

```python
params = dict(penalty=penalty, C=c_values, solver=solver)
```

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv = StratifiedKFold()
```

```python
grid = GridSearchCV(estimator = model, param_grid = params, cv=cv, scoring="accuracy",n_jobs=-1)#“Grid Search Cross-Validation” demektir. Modelin farklı hiperparametre kombinasyonlarını dener ve en iyi sonucu veren kombinasyonu bulur.
#estimator = model Hangi modeli optimize edeceğini söylüyorsun.
#param_grid = params Denenecek parametrelerin listesini/dictionarysini verir.
#cv=cv her kombinasyonu 5 katlı cross-validation ile test eder.
#scoring="accuracy" Modeli değerlendirirken hangi metriğin kullanılacağını belirler.
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: \n ", confusion_matrix(y_pred, y_test))
```

```python
# one vs rest
```

```python
from sklearn.multiclass import OneVsOneClassifier, OneVsRestClassifier
```

```python
onevsonemodel = OneVsOneClassifier(LogisticRegression())
```

```python
onevsrestmodel = OneVsRestClassifier(LogisticRegression())
```

```python
onevsonemodel.fit(X_train, y_train)
y_pred = onevsonemodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: \n ", confusion_matrix(y_pred, y_test))
```

```python
onevsrestmodel.fit(X_train, y_train)
y_pred = onevsrestmodel.predict(X_test)
score = accuracy_score(y_pred, y_test)
print("score: ", score)
print(classification_report(y_pred, y_test))
print("confusion matrix: \n ", confusion_matrix(y_pred, y_test))
```
