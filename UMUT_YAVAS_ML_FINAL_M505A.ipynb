# 3. DATA EXPLORATION
# Dataset Overview
The dataset contains **3,276 water samples** with **10 features** (9 input features and 1 target variable). Each sample represents water quality measurements from different sources.


## Water Quality Parameters
###Features:
- **pH Level:** Measures acidity on a scale of 0-14, with 6.5-8.5 considered safe for drinking
- **Hardness:** Concentration of calcium and magnesium salts measured in mg/L
- **Total Dissolved Solids (TDS):** Concentration of dissolved substances in water measured in ppm
- **Chloramines:** Disinfectants used to control bacteria, formed by combining chlorine and ammonia
- **Sulfate:** Naturally occurring mineral that can affect taste at high concentrations
- **Conductivity:** Water's ability to conduct electricity, indicating ion concentration
- **Organic Carbon:** Amount of organic matter that can serve as food for microorganisms
- **Trihalomethanes:** Compounds formed when chlorine reacts with organic matter
- **Turbidity:** Cloudiness caused by suspended particles measured in NTU

### Target Variable
- **Potability:** Binary indicator where 1 = safe for drinking, 0 = not safe for drinking

## Data Dictionary:
 **pH:** Acidity level (0-14 scale)

 **Hardness:** Calcium and magnesium concentration (mg/L)

 **Solids:** Total dissolved solids (ppm)

 **Chloramines:** Disinfectant concentration (ppm)

 **Sulfate:** Sulfate mineral concentration (mg/L)

 **Conductivity:** Electrical conductivity (μS/cm)

 **Organic_carbon:** Organic matter content (mg/L)

 **Trihalomethanes:** THM compounds concentration (μg/L)

 **Turbidity:** Water clarity measure (NTU)

 **Potability:** Target variable (0 or 1)

#4. TRAIN-TEST SPLIT

The dataset is split into training and testing sets before any preprocessing to prevent data leakage. This ensures that information from the test set does not influence the training process.


```python
df_train, df_test = sklearn.model_selection.train_test_split(df, random_state=15)

print("df size:", df.shape)
print("df_train size:", df_train.shape)
print("df_test size:", df_test.shape)
```

# 5. Missing Value Analysis and Data Type Inspection
###Data Type Inspection
Before handling missing values, we examine the data types of each column to ensure they are appropriate for analysis. All features are numeric (float64) and the target variable Potability is integer (int64), which is suitable for classification tasks. No data type conversion is needed.

###Missing Value Analysis
After splitting the data, we check for missing values in both training and test sets. The analysis reveals that three columns contain null values:

The ph column has 368 missing values in training set, 123 missing values in test set.

The Sulfate column has 592 missing values in training set, 189 missing values in test set

Trihalomethanes column has 127 missing values in training set, 35 missing values in test set

The rest of the columns do not have missing values.

```python
display(df_train.info()), display(df_test.info())
```

```python
display(df_train.isnull().sum()), display(df_test.isnull().sum())
```

#6. Handling Missing Values
To address these missing values without causing data leakage, we calculate the median for each column using only the training set. The median is chosen over the mean because it is less sensitive to outliers. These median values are then used to fill missing entries in both training and test sets. This approach prevents data leakage issues and fills nan values with logical numbers in both train and test set.

### Verification
After filling, we verify that no missing values remain in either dataset. Both training and test sets now have complete data across all columns, ready for further analysis and model development.

```python
median_ph = df_train['ph'].median()
median_sulfate = df_train['Sulfate'].median()
median_tri = df_train['Trihalomethanes'].median()

df_train['ph'] = df_train['ph'].fillna(median_ph)
df_train['Sulfate'] = df_train['Sulfate'].fillna(median_sulfate)
df_train['Trihalomethanes'] = df_train['Trihalomethanes'].fillna(median_tri)

df_test['ph'] = df_test['ph'].fillna(median_ph)
df_test['Sulfate'] = df_test['Sulfate'].fillna(median_sulfate)
df_test['Trihalomethanes'] = df_test['Trihalomethanes'].fillna(median_tri)
```

```python
display(df_train.isnull().sum()), display(df_test.isnull().sum())
```

```python
display(df_train.describe()) , display(df_test.describe())
```

```python
display(df_train.Potability.value_counts()), display(df_test.Potability.value_counts())
```

#7. Feature Selection and Scaling

### Separating Features and Target Variable
Before model training, I separated the features from the target variable. The target variable "Potability" is removed from the feature sets and stored separately for both training and testing data.

- **X_train:** Training features (2457 samples, 9 parameters)
- **X_test:** Testing features (819 samples, 9 parameters)
- **y_train:** Training target values (2457 samples)
- **y_test:** Testing target values (819 samples)
###Feature Scaling
 I applied StandardScaler to standardize the features The scaler is fitted only on the training data to prevent data leakage, then used to transform both
training and test sets.


```python
X_train = df_train.drop("Potability", axis=1)
X_test = df_test.drop("Potability", axis=1)
y_train = df_train["Potability"]
y_test = df_test["Potability"]

print("x_train size:", X_train.shape)
print("x_test size:", X_test.shape)
print("y_train size:", y_train.shape)
print("y_test size:", y_test.shape)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

#8. MODEL TRAING
For model development, it is selected four different classification algorithms to predict water potability. Each model was tuned using cross-validation (5-fold) to find the best hyperparameters. The training was performed on the scaled feature set (X_train_scaled) with the target variable (y_train).

#Preferred models:
**1. K-Nearest Neighbors – Hyperparameter Tuning with GridSearchCV**

**2. Decision Tree – Hyperparameter Tuning with GridSearchCV**

**3. Logistic Regression – Hyperparameter Tuning with GridSearchCV**

**4. Random Forest – Hyperparameter Tuning with RandomizedSearchCV**

##8.1. K-NEAREST NEIGHBORS CLASSIFIER
 I performed hyperparameter tuning using GridSearchCV with 5-fold cross-validation to find the optimal configuration.
Best Parameters: {'n_neighbors': 9, 'weights': 'distance'}
Best Cross-Validation Score: 0.6455

The optimal configuration uses 9 neighbors with distance-based weighting, meaning closer neighbors have a greater influence on the prediction than those farther away. This approach achieved a cross-validation accuracy of 64.55%.

```python
knn_params = {
    'n_neighbors': [3, 5, 7, 9],
    'weights': ['uniform', 'distance']
}
```

```python
knn_grid = GridSearchCV(estimator = KNeighborsClassifier(), param_grid=knn_params, cv=5,scoring="accuracy")
```

```python
knn_grid.fit(X_train_scaled,y_train)
```

```python
knn_grid.best_params_
```

```python
knn_grid.best_score_
```

##8.2. DECISION TREES
I performed hyperparameter tuning using GridSearchCV with 5-fold cross-validation to find the optimal configuration.
Best Parameters: {'criterion': 'gini', 'max_depth': 5, 'max_features': None, 'splitter': 'best'}

Best Cross-Validation Score: 0.6435

The optimal configuration uses a maximum depth of 5 with gini impurity as the splitting criterion, considering all features at each split. This approach achieved a cross-validation accuracy of 64.35%.

```python
dt_param = {
    "criterion" : ["gini", "entropy", "log_loss"],
    "splitter" : ["best", "random"],
    "max_depth" : [1,2,3,4,5,15,None],
    "max_features" : ["sqrt", "log2", None]
}
```

```python
dt_grid = GridSearchCV(estimator = DecisionTreeClassifier(random_state=42), param_grid=dt_param, cv=5,scoring="accuracy")
dt_grid.fit(X_train_scaled, y_train)
```

```python
dt_grid.best_params_
```

```python
dt_grid.best_score_
```

##8.3. LOGISTIC REGRESSION-->LOWEST CV SCORE!
I performed hyperparameter tuning using GridSearchCV with 5-fold cross-validation to find the optimal configuration.
Best Parameters: {'C': 0.1, 'penalty': 'l2', 'solver': 'liblinear'}

Best Cross-Validation Score: 0.6247

The optimal configuration uses a regularization strength of 0.1 with L2 penalty, optimized using the liblinear solver. This approach achieved a cross-validation accuracy of 62.47%.

```python
logistic_params = {
    'C': [0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']
}
```

```python
logistic_grid = GridSearchCV(estimator = LogisticRegression(), param_grid=logistic_params, cv=5,scoring="accuracy")
logistic_grid.fit(X_train_scaled, y_train)
```

```python
logistic_grid.best_params_
```

```python
logistic_grid.best_score_
```

##8.4. RANDOM FOREST --> HIGHEST CV SCORE!

 Given the larger hyperparameter space, I used RandomizedSearchCV with 5-fold cross-validation to efficiently find the optimal configuration.

Best Parameters: {'n_estimators': 500, 'min_samples_split': 3, 'max_features': None, 'max_depth': 10}

Best Cross-Validation Score: 0.6724

The optimal configuration uses 500 trees with a maximum depth of 10, considering all features at each split, and requiring a minimum of 3 samples to split an internal node. This approach achieved the highest cross-validation accuracy among all models at 67.24%.

```python
forest_Classsifier = RandomForestClassifier(random_state=15)
```

```python
rf_params = {
    "max_depth": [5, 10, 15, None],
    "max_features": [5, 8, 'log2', None],
    "min_samples_split": [3, 10, 15],
    "n_estimators": [100, 200, 500]
}
```

```python
rscv = RandomizedSearchCV(estimator=forest_Classsifier, param_distributions=rf_params,n_iter=10,cv=5, random_state=42, n_jobs=-1)
rscv.fit(X_train_scaled, y_train)
```

```python
rscv.best_params_
```

```python
rscv.best_score_
```

#9. MODEL EVALUATION – RANDOM FOREST (BEST MODEL)

After comparing all four models, Random Forest achieved the highest cross-validation score (0.6724). We selected it as our final model and evaluated its performance on the unseen test set.

The model achieved a test accuracy of 0.6703, which is very close to the cross-validation score, indicating good generalization without overfitting.

The confusion matrix shows that the model correctly identified 445 non-potable samples and 104 potable samples, while misclassifying 26 non-potable and 244 potable samples.

The classification report reveals that the model performs very well on class 0 (non-potable) with 94% recall, but struggles with class 1 (potable) achieving only 30% recall. This suggests the model is better at detecting unsafe water than confirming safe water.

```python
best_rf = rscv.best_estimator_
```

```python
y_pred_RF = best_rf.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print(accuracy_score(y_test, y_pred_RF))
print(confusion_matrix(y_test, y_pred_RF))
print(classification_report(y_test, y_pred_RF))
```
